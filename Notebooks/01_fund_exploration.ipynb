## AMFI Code Understanding

AMFI (Association of Mutual Funds in India) assigns a unique AMFI code to every mutual fund scheme.

The AMFI code acts as a unique identifier for each scheme and is used to connect information across multiple datasets.

In this project, the AMFI code is present in:
- fund_master.csv
- nav_history.csv
- scheme_performance.csv
- investor_transactions.csv
- portfolio_holdings.csv

Since the same AMFI code appears in multiple datasets, it can be used as the primary key for merging and analyzing data.

```python
# AMFI Code Validation
fund_master_codes = set(fund_master["amfi_code"])
nav_history_codes = set(nav_history["amfi_code"])

missing_codes = fund_master_codes - nav_history_codes

print("Missing AMFI Codes:", missing_codes)

if len(missing_codes) == 0:
    print("AMFI Code Validation Successful!")
    print("All AMFI codes in fund_master are present in nav_history.")
else:
    print("AMFI Code Validation Failed!")
```
